```python
import pandas as pd
from ofxparse import OfxParser

# Load the QFX file
with open("sample.qfx") as f:
    ofx = OfxParser.parse(f)

transactions = []
for txn in ofx.account.statement.transactions:
    transactions.append({
        "Date": txn.date.strftime("%Y-%m-%d"),
        "Type": txn.type,
        "Amount": txn.amount,
        "Payee": txn.payee,
        "Memo": txn.memo
    })

# Convert to DataFrame
df = pd.DataFrame(transactions)

# Save to Excel
df.to_excel("quicken_to_excel.xlsx", index=False)

print("✅ Conversion complete! File saved as quicken_to_excel.xlsx")
```
